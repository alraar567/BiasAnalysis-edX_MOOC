```python
prob_list = problem_exam['ID']
prob_list
```

```python
grades_normalized = pd.DataFrame(index = data['user_id'].unique(), columns=prob_list)
for i in prob_list:
  it = 'i4x://UC3Mx/xxxxx/problem/'+ i
  users_id = list(data['user_id'].loc[(data['event_type'] == 'problem_check') & (data['element_id']==it)])
  #print('problem:',i)
  grades = list(data['grade'].loc[(data['event_type'] == 'problem_check') & (data['element_id']==it)])
  max_grades = list(data['max_grade'].loc[(data['event_type'] == 'problem_check') & (data['element_id']==it)])
  if len(users_id) == len(grades) == len(max_grades):
    for j in range(0,len(users_id)):
        grades_normalized.loc[users_id[j],i] = grades[j]/max_grades[j]
grades_all = grades_normalized.fillna('not attempted')
grades_one = grades_all.loc[(grades_all!='not attempted').any(axis=1)]
grades = grades_one.replace('not attempted',0)
grades
```

```python
grades_weighted = pd.DataFrame(index = data['user_id'].unique(), columns=prob_list)
for i in prob_list:
  it = 'i4x://UC3Mx/xxxx/problem/'+ i
  users_id = list(data['user_id'].loc[(data['event_type'] == 'problem_check') & (data['element_id']==it)])
  #print('problem:',i)
  grades = list(data['grade'].loc[(data['event_type'] == 'problem_check') & (data['element_id']==it)])
  max_grades = list(data['max_grade'].loc[(data['event_type'] == 'problem_check') & (data['element_id']==it)])
  if len(users_id) == len(grades) == len(max_grades):
    for j in range(0,len(users_id)):
        grades_weighted.loc[users_id[j],i] = (grades[j]/max_grades[j]) * int(problem_exam['Weight'].loc[problem_exam['ID']==i])
gr_all = grades_weighted.fillna('not attempted')
gr_one = gr_all.loc[(gr_all!='not attempted').any(axis=1)]
gr = gr_one.replace('not attempted',0)
gr
```
