```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
indicadores = pd.read_csv('./Indicadores/indicadores_all.csv',index_col=0)
indicadores
```

```python
indicadores['First attempt correct nongraded (%)'] = indicadores['First attempt correct nongraded (%)'].replace('0 problems attempted',0)
indicadores['abandono'] = indicadores['abandono'].replace('No',0).replace('Sí',1)
indicadores['curso_superado'] = indicadores['curso_superado'].replace('No',0).replace('Sí',1)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import roc_curve, roc_auc_score, auc, average_precision_score, precision_recall_curve
from matplotlib import pyplot as plt
from sklearn.metrics import mean_squared_error
```

# **Abandono**

```python
X = indicadores.drop(['abandono'],axis=1)
y = indicadores['abandono']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
# Linear Regression model: Para predecir el abandono.
reg = LinearRegression().fit(X_train, y_train)
print('score:',reg.score(X_train, y_train))
print('coef:',reg.coef_)
print('intercept:',reg.intercept_)
print('test score:',reg.score(X_test,y_test))
```

```python
1-mean_squared_error(y_test, reg.predict(X_test))
```

```python
predictions = reg.predict(X_test)
print(roc_auc_score(y_test, reg.predict(X_test)))

fpr, tpr, _ = roc_curve(y_test, predictions)

#plt.reg()
plt.plot(fpr, tpr)
plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('ROC curve')
plt.show()
```

```python
# Decision Tree model: Para predecir el abandono.
clf = DecisionTreeClassifier(random_state=0).fit(X_train, y_train)
print('score:',clf.score(X_train, y_train))
print('test score:',clf.score(X_test,y_test))
```

```python
1-mean_squared_error(y_test, clf.predict(X_test))
```

```python
predictions = clf.predict_proba(X_test)
print(roc_auc_score(y_test, clf.predict(X_test)))

fpr, tpr, _ = roc_curve(y_test, predictions[:,1])

plt.clf()
plt.plot(fpr, tpr)
plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('ROC curve')
plt.show()
```

```python
fpr, tpr, _ = precision_recall_curve(y_test, clf.predict(X_test))

plt.plot(fpr, tpr)
plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('ROC curve')
plt.show()
```

```python
# Random Forest model: Para predecir el abandono.
clf2 = RandomForestClassifier(max_depth=2, random_state=0).fit(X_train, y_train)
print('score:',clf2.score(X_train, y_train))
print('test score:',clf2.score(X_test,y_test))
```

```python
1-mean_squared_error(y_test, clf2.predict(X_test))
```

```python
average_precision_score(y_test, clf2.predict(X_test))
```

```python
precision_recall_curve(y_test, clf2.predict(X_test))
```

```python
fpr, tpr, _ = precision_recall_curve(y_test, clf2.predict(X_test))

plt.plot(fpr, tpr)
plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('ROC curve')
plt.show()
```

```python
predictions = clf2.predict(X_test)
print(roc_auc_score(y_test, clf2.predict(X_test)))

fpr, tpr, _ = roc_curve(y_test, predictions[:,1])

plt.clf()
plt.plot(fpr, tpr)
plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('ROC curve')
plt.show()
```

**Nota final del curso**

```python
grades_per_exam = pd.read_csv('./Grades/grades_per_exam.csv',index_col=0)
grades_per_exam
```

```python
indicadores = indicadores.join(grades_per_exam['Course grade'])
indicadores = indicadores.fillna(0)
indicadores['Course grade'].unique()
```

```python
X = indicadores.drop(['Course grade'],axis=1)
y = indicadores['Course grade']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
```

```python
# Linear Regression model: Para predecir la nota final.
reg = LinearRegression().fit(X_train, y_train)
print('score:',reg.score(X_train, y_train))
print('coef:',reg.coef_)
print('intercept:',reg.intercept_)
print('test score:',reg.score(X_test,y_test))
```

```python
predictions = reg.predict(X_test)
print(roc_auc_score(y_test, reg.predict(X_test)))

fpr, tpr, _ = roc_curve(y_test, predictions[:,1])

plt.plot(fpr, tpr)
plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('ROC curve')
plt.show()
```

```python
y_test.unique()
```

```python
# Decision Tree model: Para predecir la nota final.
regressor = DecisionTreeRegressor(random_state=0).fit(X_train, y_train)
print('score:',regressor.score(X_train, y_train))
print('test score:',regressor.score(X_test,y_test))
```

```python
predictions = regressor.predict(X_test)
roc_auc_score(y_test, regressor.predict(X_test), multi_class='ovo')

fpr, tpr, _ = roc_curve(y_test*100, predictions[:,1]*100)

plt.plot(fpr, tpr)
plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('ROC curve')
plt.show()
```

```python
# Random Forest model: Para predecir la nota final.
regressor2 = RandomForestRegressor(max_depth=2, random_state=0).fit(X_train, y_train)
print('score:',regressor2.score(X_train, y_train))
print('test score:',regressor2.score(X_test,y_test))
```
