```python
data2015 = data.loc[data['time'].str.contains('2015-')]
data2015
```

```python
import datetime

date_format = "%Y-%m-%d"
d1 = datetime.datetime.strptime("2015-04-28", date_format).date()
d2 = datetime.datetime.strptime("2015-06-16", date_format).date()
step = datetime.timedelta(days=7)
weeks = []
while d1 <= d2:
    print(d1.strftime(date_format))
    weeks.append(d1.strftime(date_format))
    d1 += step
```

```python
for w in range(0,len(weeks[:-1])):
  print(1,w+1)
```

```python
date_format = "%Y-%m-%d"
from datetime import timedelta
for w in range(0,len(weeks[:-1])):
  print('week:',w)
  d1 = datetime.datetime.strptime(weeks[0], date_format).date()
  d2 = datetime.datetime.strptime(weeks[w+1], date_format).date()- timedelta(days=1)
  step = datetime.timedelta(days=1)
  print('d1:',d1)
  df = data2015.loc[data['time'].str.contains(str(d1))]
  while d1 < d2:
      d1 += step
      df = pd.concat([df,data2015.loc[data['time'].str.contains(str(d1))]])
  print(len(df))
  print('d2:',d2)
  df.to_csv('./Modelos/Cummulative/Weeks/weeks{}.csv'.format(str(1)+'_'+str(w+1)))
```
