# **Acumulativo**


```python
# Problemas por semana
prob_w1 = list(problems.loc[problems['Name'].str.contains('Exam 1')].index)
prob_w2 = list(problems.loc[problems['Name'].str.contains('Exam 1')].index)
prob_w3 = list(problems.loc[(problems['Name'].str.contains('Exam 1')) | (problems['Name'].str.contains('Exam 2'))].index)
prob_w4 = list(problems.loc[(problems['Name'].str.contains('Exam 1')) | (problems['Name'].str.contains('Exam 2')) | (problems['Name'].str.contains('Exam 3'))].index)
prob_w5 = list(problems.loc[(problems['Name'].str.contains('Exam 1')) | (problems['Name'].str.contains('Exam 2')) | (problems['Name'].str.contains('Exam 3')) | (problems['Name'].str.contains('Exam 4'))].index)
prob_w6 = list(problems.index)
prob_w7 = list(problems.index)
probs = [prob_w1,prob_w2,prob_w3,prob_w4,prob_w5,prob_w6,prob_w7]

# Users from configuration f
users_f = list(conf_all_students.index)
```

```python
for w in range(1,7):
  print('\nWeek:',w)
  week_data = pd.read_csv('./Modelos/Cummulative/Weeks/weeks1_{}.csv'.format(w),index_col=0)
  week_data.drop(['msg_parent'],axis=1)
  grades = pd.DataFrame(index=users_f,columns=probs[w-1])
  for p in probs[w-1]:
    print('Problem:',p)
    users_id = list(week_data['user_id'].loc[(week_data['event_type'] == 'problem_check') & (week_data['element_id'].str.contains(p))])
    gr = list(week_data['grade'].loc[(week_data['event_type'] == 'problem_check') & (week_data['element_id'].str.contains(p))])
    max_gr = list(week_data['max_grade'].loc[(week_data['event_type'] == 'problem_check') & (week_data['element_id'].str.contains(p))])
    if len(users_id) == len(gr) == len(max_gr):
      for u in range(0,len(users_id)):
        if users_id[u] in users_f:
            grades.loc[users_id[u],p] = gr[u]/max_gr[u]
  grades = grades.fillna(0)
  grades.to_csv('./Modelos/Cummulative/Configuración all students/grades_allatt_week{}.csv'.format(w),header=True)
```
