```python
problem_exam = pd.read_csv('./problem_exam.csv')
problem_exam
```

```python
prob_list = problem_exam['ID']
prob_list
```

```python
for i in range(0,len(prob_list)):
  it = 'i4x://UC3Mx/xxxx/problem/'+ prob_list[i]
  max_grade = list(data['max_grade'].loc[(data['element_id'] == it) & (data['event_type'] == 'problem_check')])[0]
  problem_exam.loc[i,'Maximum grade'] = max_grade
problem_exam
```

```python
problem_exam.to_csv('./problems.csv',header=True,index = False)
```

```python
users_problems_simple = pd.DataFrame(index = data['user_id'].unique(), columns=prob_list)
users_problems_simple
```

```python
for i in prob_list:
  it = 'i4x://UC3Mx/xxxx/problem/'+ i
  users_id = list(data['user_id'].loc[(data['event_type'] == 'problem_check') & (data['element_id']==it)])
  print('problem:',i)
  grades = list(data['grade'].loc[(data['event_type'] == 'problem_check') & (data['element_id']==it)])
  if len(users_id) == len(grades):
    for j in range(0,len(users_id)):
        users_problems_simple.loc[users_id[j],i] = grades[j]
```

```python
users_problems_simple = users_problems_simple.fillna(0)
users_problems_simple
```
