```python
# Conf a: Haber hecho al menos un ejercicio
conf_a = indicadores.loc[indicadores['n_attempted'] >=1]
conf_a
```

```python
conf_a_g = grades_all.loc[conf_a.index]
conf_a_g
```

```python
# Conf b: Haber abierto al menos un video
conf_b = indicadores_all.loc[indicadores_all['n_videos_opened'] >=1]
conf_b
```

```python
conf_b_g = grades_all.loc[conf_b.index]
conf_b_g
```

```python
# Conf c: Haber visto al menos un video
conf_c = indicadores_all.loc[indicadores_all['n_videos_watched'] >=1]
conf_c
```

```python
conf_c_g = grades_all.loc[conf_c.index]
conf_c_g
```

```python
# Conf d: Haber hecho al menos un ejercicio y abierto al menos un video
conf_d = indicadores_all.loc[(indicadores_all['n_attempted'] >=1) & (indicadores_all['n_videos_opened'] >= 1)]
conf_d
```

```python
conf_d_g = grades_all.loc[conf_d.index]
conf_d_g
```

```python
# Conf e: Haber participado en el foro al menos una vez
conf_e = indicadores.loc[(indicadores['n_threads'] >=1) | (indicadores['n_responses'] >=1) | (indicadores['n_comments'] >=1)]
conf_e
```

```python
conf_e_g = grades_all.loc[conf_e.index]
conf_e_g
```

```python
# Conf f: Haber intentado al menos un problema en dos exámenes diferentes
n_problems_exam = pd.DataFrame(index = data['user_id'].unique(), columns = ['Exam 1','Exam 2','Exam 3','Exam 4','Exam 5'])

for e in n_problems_exam.columns:
  print(e)
  probs = list(problems.loc[problems['Name'].str.contains(e)].index)
  for u in n_problems_exam.index:
    n_problems_exam.loc[u,e] = 0
    for p in probs:
      if grades_all.loc[u,p] != 'not attempted':
        n_problems_exam.loc[u,e] += 1

n_problems_exam['more_than_2'] = 0
for u in n_problems_exam.index:
  for e in n_problems_exam.columns[:-1]:
    if n_problems_exam.loc[u,e] >= 1:
      n_problems_exam.loc[u,'more_than_2'] += 1

conf_f = n_problems_exam.loc[n_problems_exam['more_than_2'] >=2]
conf_f
```

```python
conf_f_g = grades_all.loc[conf_f.index]
conf_f_g
```

```python
# Conf g: Haber hecho al menos un ejercicio y visto al menos un video
conf_g = indicadores_all.loc[(indicadores_all['n_attempted'] >=1) & (indicadores_all['n_videos_watched'] >= 1)]
conf_g
```

```python
conf_g_g = grades_all.loc[conf_g.index]
conf_g_g
```
