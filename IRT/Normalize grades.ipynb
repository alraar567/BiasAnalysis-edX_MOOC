```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data_name = './datos_hkust_all.csv'
```

```python
data = pd.read_csv(data_name)
```

```python
problem_exam = pd.read_csv('./old files/problem_exam.csv')
problem_exam
```

```python
prob_list = problem_exam['ID']
prob_list
```

```python
grades_normalize = pd.DataFrame(index = data['user_id'].unique(), columns=prob_list)
grades_normalize
```

```python
for i in prob_list:
  it = 'i4x://UC3Mx/IT.1.1x/problem/'+ i
  users_id = list(data['user_id'].loc[(data['event_type'] == 'problem_check') & (data['element_id']==it)])
  print('problem:',i)

  grades = list(data['grade'].loc[(data['event_type'] == 'problem_check') & (data['element_id']==it)])
  max_grades = list(data['max_grade'].loc[(data['event_type'] == 'problem_check') & (data['element_id']==it)])

  if len(users_id) == len(grades) == len(max_grades):
    for j in range(0,len(users_id)):
        grades_normalize.loc[users_id[j],i] = grades[j]/max_grades[j]
```

```python
grades_normalize = grades_normalize.fillna('not attempted')
for i in prob_list:
  print(grades_normalize[i].unique())
```

```python
grades_normalize.to_csv('./grades_normalize_all.csv',header=True)
```

```python
grades_normalize_one_attempted = grades_normalize.loc[(grades_normalize!='not attempted').any(axis=1)]
grades_normalize_one_attempted
```

```python
grades_normalize_one_attempted.to_csv('./grades_normalize_one_attempted.csv',header=True)
```

```python
grades_normalize_all_attempted = grades_normalize.loc[(grades_normalize!='not attempted').all(axis=1)]
grades_normalize_all_attempted
```

```python
grades_normalize_all_attempted.to_csv('./grades_normalize_all_attempted.csv',header=True)
```
