```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data_name = './datos_hkust_all.csv'
```

```python
data = pd.read_csv(data_name)
```

```python
indicadores0 = pd.read_csv('./Indicadores/indicadores0.csv',index_col=0)
indicadores0
```

```python
indicadores1 = pd.read_csv('./Indicadores/indicadores1.csv',index_col=0)
indicadores1
```

```python
videos_percents = pd.read_csv('./Indicadores/videos_percents_oo1.csv',index_col=0)
videos_percents
```

```python
videos_percents_total = pd.DataFrame(index = data['user_id'].unique(), columns=['videos visualized (%)'])
videos_percents_total['videos visualized (%)'] = np.mean(videos_percents,axis=1)
videos_percents_total
```

```python
indicadores_all = indicadores0.drop(['List of grades'],axis=1)
indicadores_all = indicadores_all.join(indicadores1.drop(['Total number of attempts','Average number of attempts'],axis=1))
indicadores_all = indicadores_all.join(videos_percents_total)
indicadores_all
```

```python
abandono = pd.read_csv('./Abandono/abandono.csv',index_col=0)
abandono
```

```python
indicadores_all = indicadores_all.join(abandono.drop(['Exam1','Exam2','Exam3','Exam4','Exam5'],axis=1))
indicadores_all
```

```python
indicadores_all.to_csv('./Indicadores/indicadores_all.csv',header=True)
```
