avg_pauses: Average number of times a student pauses the videos

```python
videos = pd.read_csv('./Indicadores/videos.csv')
videos['Seconds list'] = [np.zeros((int(i))) for i in videos['Duration']]
vid_list = videos['ID']
```

```python
total_pauses = pd.DataFrame(index = data['user_id'].unique(), columns=vid_list)
total_pauses = total_pauses.fillna(0)
for v in vid_list:
  users = list(data['user_id'].loc[(data['event_type']=='pause_video')&(data['element_id'].str.contains(v))])
  unique_users = np.unique(np.array(users))
  for u in unique_users:
    total_pauses.loc[u,v] += 1
total_pauses
```

```python
np.mean(total_pauses,axis=1)
```

percentage_attempted: Percentage of attempted exercises (with respect to the total number of exercises)

```python
prob_list = grades_nongraded.columns
prob_list
```

```python
attempted_exercises = pd.DataFrame(index = data['user_id'].unique(), columns=prob_list)
attempted_exercises = attempted_exercises.fillna(0)
for p in prob_list:
  print(p)
  users = list(data['user_id'].loc[(data['event_type']=='problem_check')&(data['element_id'].str.contains(p))])
  unique_users = np.unique(np.array(users))
  for u in unique_users:
    attempted_exercises.loc[u,p] = 1
attempted_exercises
```

```python
np.sum(attempted_exercises,axis=1)/len(prob_list)/100
```
