```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
## Criterios de abandono:
# - abandono: No hace los últimos 2 exámenes o el último examen y suspende (porque puede haber alumnos que tras los primeros tres exámenes aprueben el curso, y paren de hacer actividades)
```

```python
grades_per_exam = pd.read_csv('./Grades/grades_per_exam.csv',index_col=0)
grades_per_exam
```

```python
abandono = pd.DataFrame(index = grades_per_exam.index, columns=['Exam1','Exam2','Exam3','Exam4','Exam5','abandono','curso_superado'])
abandono
```

```python
for user in grades_per_exam.index:
  for i in range(1,6):
    ex = grades_per_exam.loc[user,'EXAM {}'.format(i)]
    if ex == 0:
      abandono.loc[user,'Exam{}'.format(i)] = 'Sí'
    else:
      abandono.loc[user,'Exam{}'.format(i)] = 'No'
  gr = grades_per_exam.loc[user,'Course grade']
  if gr < 0.5:
    abandono.loc[user,'curso_superado'] = 'No'
  else:
    abandono.loc[user,'curso_superado'] = 'Sí'
  if (abandono.loc[user,'Exam4'] == 'Sí' and abandono.loc[user,'Exam5'] == 'Sí') or abandono.loc[user,'Exam5'] == 'Sí':
    if abandono.loc[user,'curso_superado'] == 'No':
      abandono.loc[user,'abandono'] = 'Sí'
    else:
      abandono.loc[user,'abandono'] = 'No'
  else:
    abandono.loc[user,'abandono'] = 'No'

abandono.iloc[:,0:5] = abandono.iloc[:,0:5].fillna('Sí')
abandono['curso_superado'] = abandono['curso_superado'].fillna('No')
abandono['abandono'] = abandono['abandono'].fillna('Sí')
abandono
```
